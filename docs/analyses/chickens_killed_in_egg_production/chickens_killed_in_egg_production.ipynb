# How many chickens are killed in egg production?

[88 billion land animals](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat) were killed in 2024 to produce meat worldwide.
[Almost 90%](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat?stackMode=relative) of those land animals were chicken.

Those are the estimates provided by FAOSTAT, the statistical database of the Food and Agriculture Organization of the United Nations.
But FAOSTAT does not provide an analogous estimate for the number of chickens killed to produce eggs.
And, as we will see later, that number is estimated to be surprisingly large.

This technical publication is a companion to the article ***.
Here we explain how to estimate the toll of chickens killed to produce eggs, based on different pieces of information from FAOSTAT and other sources.

## FAOSTAT's crops and livestock products

Every year, FAOSTAT gathers crops and livestock data from each country's official statistics.
FAOSTAT harmonizes the data to ensure figures are consistent with one another and use the same units.
They then update the content of their dataset called [Production: Crops and livestock products](https://www.fao.org/faostat/en/#data/QCL).
From now on, we'll simply call this dataset "QCL", which is the code FAOSTAT uses to identify it.

Unfortunately, not every country reports all of its data every year.
When official data are missing, FAOSTAT may take values from another organization, or estimate them.
To be transparent about the provenance of each data point, they publish each value with a "flag".
This flag clarifies whether a given value is official, taken from another organization, estimated or imputed by FAOSTAT, or missing.

The QCL dataset is indexed by country, year, element (roughly speaking, the metric), item (roughly, the food product), and unit.
For example, the number of chickens killed for meat corresponds to element "Producing Animals/Slaughtered", and item "Meat of chickens, fresh or chilled".
For World in year 2024, this indicator gives 78533923 thousand animals (so, close to 79 billion chickens).
This number is currently flagged as "Estimated value", and it may change in future dataset updates.

To understand the content of the dataset, these are the main sources of information to consider:
- [The QCL methodology](https://files-faostat.fao.org/production/QCL/QCL_methodology_e.pdf), which is a short document that explains the main concepts.
- [The QCL metadata](https://www.fao.org/faostat/en/#data/QCL/metadata), specifically section 5: Statistical presentation.
- [The QCL definitions and standards](https://www.fao.org/faostat/en/#data/QCL) (which can be accessed from the main QCL data site by clicking on "Definitions and standards"), specifically for "Element", "Item", and "Glossary".
- [The QCL revision note](https://files-faostat.fao.org/production/QCL/Q_Revision_Note_e.pdf), which is a short document explaining a methodological revision carried out in 2016.
- [The QCL hen egg technical conversion factors](https://files-faostat.fao.org/production/QCL/Hen_Eggs_technical_conversion_factor.pdf), which contains a table with an average egg weight used for certain countries when no other data is available.

The QCL definitions and standards' glossary distinguishes livestock products based on whether they are primary or processed, and whether they come from live or slaughtered animals:

| | From live animals | From slaughtered animals |
|---|---|---|
| **Primary products** | Include milk, eggs, honey, beeswax, fibres of animal origin | Include meat, edible offal, slaughter fats, fresh hides and skins |
| **Processed products** | Include dairy products like butter, cheese, dried eggs | Include bacon, ham, sausages, canned meat, lard, tallow |

Eggs are therefore a primary product of live animals, and chicken meat is a primary product of slaughtered animals.
There are no processed egg or chicken products in QCL.

In the QCL methodology there are three sections that are relevant for our analysis: livestock numbers, meat, and eggs.

### Livestock numbers

According to the methodology document, their livestock numbers are meant to cover all domestic animals irrespective of their age and the place or purpose of their breeding. However, they point out two caveats:
> In certain countries, data for chickens, ducks and turkeys do not yet seem to represent the total number of these birds.
    
This could mean that some countries count only commercial farms and may undercount their backyard birds.
    
> Certain other countries give a single figure for all poultry; data for these countries are shown under 'Chickens'.
    
This means that for some countries, the number of chickens may include other birds.
This ambiguity may occur even where figures are flagged as official.

The QCL definitions and standards' elements include the following two definitions for livestock numbers:
- Stocks: this element gives the total number of live animals at a given time.
> This variable indicates the number of animals of the species present in the country at the time of enumeration. It includes animals raised either for draft purposes or for meat, eggs and dairy production or kept for breeding. [...] The enumeration to be chosen, when more than one survey is taken, is the closest to the beginning of the calendar year.

- Laying: this element gives the number of birds that lay eggs. However, no specific definition is provided for this element. As we explain in [a later section](#Does-FAOSTAT-measure-hens-at-a-point-in-time-or-the-cumulative-number-of-hens), there is an important ambiguity in how live hens are counted.

### Meat
According to the methodology document, meat is counted where the animal is slaughtered, not where it was born or raised. They mention some inconsistencies in reporting (e.g. the US includes giblets, and some countries may or may not include them). They also explain that:
> All data shown relate to total meat production, that is, from both commercial and farm slaughter. [...]

This means that the number of chickens killed for meat production includes both chickens from commercial slaughterhouses and those killed on farms. However, the QCL metadata provides a somewhat different claim:
> "Slaughtering" is measured through activity of slaughterhouses, e.g. production of marketable meat for human consumption. Estimates of ‘other slaughtering’ can be added for a more accurate picture of meat production. 

This means that the number of chickens killed for meat comes primarily from slaughterhouse records. Animals killed elsewhere, on farms or in backyards "can be added", which implies that they may be missing, or may be included as an estimate.

### Eggs

The QCL dataset includes the following elements per country and year:
- Tonnes of egg produced (element "Production", in tonnes).
- Number of eggs produced (element "Production", in thousands).
- Number of live animals (element "Stocks", in thousands).
- Number of laying animals (element "Laying", in thousands).
- Number of eggs produced per hen (element "Yield", in number per animal).

And the QCL dataset includes the following items:
- Eggs from hens (item "Hen eggs in shell, fresh").
- Eggs from other birds (item "Eggs from other birds in shell, fresh, n.e.c.").
- Eggs from all birds (item "Eggs Primary").

Some countries report only one or a few of those elements, and FAOSTAT needs to derive one based on the other.
To do that, FAOSTAT may assume an average egg weight (how many grams a typical egg weighs) and/or an average lay rate (how many eggs a typical hen lays per year).
These values are taken from official statistics where possible; otherwise FAOSTAT assumes them.
This is explained in the QCL methodology:
> Some countries have no statistics on egg production, and estimates had to be derived from such related data as chicken or total poultry numbers and reported or assumed rates of egg laying. Most of the countries that have statistics on egg production report either the total weight of eggs or the numbers of eggs produced; data on numbers have been converted into weight, using official conversion factors wherever possible.

QCL includes eggs from all types of hens, and it includes eggs for hatching (fertilized eggs that are not eaten, but hatched into chicks for both egg-laying and meat flocks), but it excludes wasted eggs that did not make it out of the farms (e.g. eggs that broke in the farm).
They include eggs from both traditional farms (small flocks of backyard hens) and modern ones (large-scale, intensive commercial poultry farms).
This is explained in the QCL definitions and standards' item for "Eggs Primary":
> [...] Egg production by type of poultry should refer to the total production of eggs in the shell by all types of hens in both the traditional sector (individually owned small flocks) and the modern sector (large-scale, intensive commercial poultry farms). Total productions include eggs for hatching but exclude waste of farms. Countries should report in terms of both numbers and weight.

Finally, the QCL definitions and standards' glossary includes a very important explanation for the term "Egg production":
> Covers all domestic birds which have contributed to egg production during the year, wherever they lay and the corresponding total production, including eggs intended to be used for hatching but excluding waste on farms.

As we explain in [the next section](#Does-FAOSTAT-measure-hens-at-a-point-in-time-or-the-cumulative-number-of-hens), this explanation carries some ambiguity.
For reasons we justify in that section, we assume that the number of domestic birds counted for egg production is the count of all laying birds at a given point in time (and not the total number of birds that have laid during the year).

### Does FAOSTAT measure hens at a point in time, or the cumulative number of hens?

The explanation of "Egg production" in the QCL glossary, quoted at the end of [the section on eggs](#Eggs), has two possible interpretations, with crucially different consequences.
"All domestic birds which have contributed to egg production during the year" can mean:
1. All individual birds that laid eggs during the year: this means that the "Laying" element gives the **cumulative** number of birds that have laid eggs in the year.
2. All kinds of domestic birds that lay eggs: in this case, the sentence says nothing about how birds are counted, and the "Laying" element would give the number of birds that lay eggs **at any given time**.

Note that each interpretation would lead to very different numbers:
Imagine a farm that produces 30,000 eggs a year with 100 hens in lay, and replaces each hen after one year, spreading the replacements evenly over the year.
Under Option 2, the "Laying" element should be 100 (with a yield of 300 eggs per hen).
Under Option 1, "Laying" should report 200 hens (with a yield of 150 eggs per hen): the 100 that were there at the beginning of the year, plus the 100 that replaced them during the year.

Based purely on semantics, Option 1 seems to be the more natural interpretation of FAOSTAT's explanation.
All domestic birds which have contributed to egg production is the cumulative number of hens that were in lay during the year.

However, Option 2 could be interpreted as a clarification that the data includes all species that lay eggs, e.g. hens, ducks, and pigeons, and not just hens.
Option 2 could also be interpreted as a clarification that the data includes breeder hens (of chickens raised both for eggs and chickens raised for meat), and not just laying hens.

Regardless of semantics, the most important question is: which interpretation is more likely the one FAOSTAT assumed?
To me, there are strong reasons to believe that Option 2 is more plausible.

Firstly, the QCL revision note clarifies that data on agricultural productivity (e.g. yield) is not collected.
FAOSTAT derives it by dividing outputs by inputs; in the case of yield per hen, it is estimated as number of eggs divided by number of hens.
Indeed, one can easily demonstrate that FAOSTAT's yield is numerically consistent with the division of egg production (the "Production" element) by the number of laying hens (the "Laying" element).
In 99% of countries and years, the two agree to within 2%; the exceptions are mostly small islands, with very few hens.


```python
# FAOSTAT's eggs per hen compared with its production (in number of eggs) divided by its laying hens, in every country
# and year.
qcl = load_table(FAOSTAT_QCL_TABLE)
yields = qcl[(qcl["item_code"].astype(str) == "00001062") & qcl["element_code"].astype(str).isin(["005313", "005413", "005513"])]
# Keep countries only: our regions have codes starting with "OWID_", FAOSTAT's regions have codes from 5000 on, and code
# 351 ("China (FAO)") adds up mainland China, Taiwan, Hong Kong and Macao, which are also listed separately.
area_code = yields["area_code"].astype(str)
yields = yields[area_code.str.isdigit() & (pd.to_numeric(area_code, errors="coerce") < 5000) & (area_code != "351")]
yields = yields.pivot_table(index=["country", "year"], columns="element_code", values="value", observed=True).dropna()
difference = (yields["005513"] / yields["005313"] / yields["005413"] - 1).abs()
assert (difference < 0.02).mean() > 0.98, "FAOSTAT's eggs per hen no longer match its production divided by its laying hens."
# Most of the exceptions are small islands, with fewer than 100,000 hens.
assert (yields[difference >= 0.02]["005313"] < 1e5).mean() > 0.5, "The exceptions are no longer mostly small islands."
```

If FAOSTAT expected the number of laying hens to be cumulative, this calculation would not be a meaningful representation of productivity.
The fact that FAOSTAT does this calculation is evidence that they assume Option 2 above.
Hence, the element "Laying" gives the number of laying birds at a point in time (and not the cumulative number of birds during the year).

Secondly, even though the QCL's definitions and standards do not provide an explicit explanation of the "Laying" element, they do provide one for the "Stocks" element.
Stocks are very clearly defined as the number of animals in the country at the time of enumeration.

## Default estimate
The main goal of this document is to estimate the number of chickens that are killed to produce eggs.
In this section we will carry out that estimate following a rather simple approach, which we will call the default estimate.

We start by estimating the number of laying hens killed every year.
For that, we use FAOSTAT's data on the number of live hens (element "Laying", item "Hen eggs in shell, fresh"): this indicator gives the number of hens alive at a given time.
FAOSTAT has no data for some countries in some years; in those cases, we repeat the country's latest value, for at most five years.
In 2024, the global number of hens alive was around 8.6 billion.

A young female chicken (a pullet) in a typical commercial system takes around 18 weeks (the rearing period) from the moment of hatching until she starts laying eggs.
Between 18 and around 80 weeks of age (the laying period), the hen lays eggs regularly.
After this period the hen is killed, as she is no longer considered productive enough.
Both the rearing and the laying period vary with the breed and the production system, and both are usually longer in traditional systems.
We will assume a laying period of 65 weeks, as explained in [the section on the laying period](#Laying-period).

Given that there are around 8.6 billion hens alive at any given time, and each of them lasts for about 65 weeks before it's slaughtered and replaced by another one, then

8.6 billion hens / laying period x ( 1 laying period / 65 week ) x ( 52 weeks / 1 year ) ~ 6.9 billion hens / year

In other words, roughly 6.9 billion hens were slaughtered worldwide in 2024.

For each female chick hatched, on average one male chick is hatched too.
This relies on the assumption that chicks are equally likely to hatch male or female, which is supported by basic biology.
As we explain in the accompanying article***, the majority of those male chicks are killed soon after hatching.
If we assume that all those chicks are culled, this leads to 6.9 billion male chicks killed in 2024.

Therefore, the global number of chickens that hatch within the egg production industry, that are eventually killed, is around 13.8 billion in 2024.

This simple approach to estimate the number of chickens killed per year can also be performed at the country level.
When we do that, we see that the countries killing the largest number of chickens in 2024 were:
- China, at 5.4 billion,
- India, at 1.1 billion,
- Indonesia, at 0.66 billion,
- United States, at 0.60 billion, and
- Brazil, at 0.44 billion.

The QCL dataset has data from 1961, but we only use data from 1990 onward.
Our assumed laying period is a reasonable choice for commercial hens, but may be short for backyard ones, which may be kept for longer.
Until the late 1980s, a large share of the world's hens were in countries where eggs came mostly from backyard flocks, so our estimated number of hens killed may be too high for those years.
[The section on the starting year](#Starting-year) explains how we chose the cutoff year of 1990.

### Limitations on the default estimate
Some factors push the previous estimate up, and others pull it down.

#### Arguments that push the estimate up
- Pre-slaughter mortality.
A fraction of pullets die before they become laying hens, and a fraction of laying hens die, or are killed, before they reach the slaughter age.
If we add these pre-slaughter deaths, with the rates that Faunalytics uses (3.5% of pullets die before they start laying, and 8% of hens during the laying period), the number goes up to about 7.2 billion hens, leading to a total of about 14.7 billion chickens killed in 2024.
This figure is consistent with [a recent publication by Faunalytics](https://faunalytics.org/global-chick-culling-2019-2024/), that puts the number of male chicks culled at 8 billion per year, since they assume a shorter laying period of 62 weeks.
- Disease outbreaks.
The estimate leaves out hens culled in disease outbreaks such as avian influenza.
- Shorter laying period.
The typical laying period among China's commercial hens could be shorter than our adopted 65 weeks, which would increase the global number of chickens killed.

#### Arguments that push the estimate down
- Backyard hens.
As we discuss in [the section on the laying period](#Laying-period), in many low- and middle-income countries, mostly in Asia and Africa, egg production is dominated by backyard hens, which may be kept for longer than our assumed 65 weeks.
If so, our estimates overstate the number of chickens killed in those countries; however, on a global scale, their contribution is small.
- Pullets.
For some countries, FAOSTAT's number of laying hens includes pullets.
For those countries, our default estimate would need to account for the additional 18 weeks of life of the hens before they start laying (so, we would need to replace our 65 by 83 weeks).
This issue is known for the UK, but it can also affect other countries.
- Breeders.
FAOSTAT's number of laying hens may include breeders for some countries.
They are also part of the production system, so it is reasonable to include them in the calculation, just like any other type of hen.
But among those hens there are also broiler breeders: hens whose fertilized eggs will become chickens used for meat.
The number of broiler breeders killed should therefore be attributed to meat production, not to egg production.
In the United States, broiler breeders are 16% of all layers (62 million of the 377 million in 2024, see [the section on the United States](#United-States)); we do not know their share in other countries.
- Inconsistent data for China.
China is the biggest egg producer, and FAOSTAT numbers of eggs and hens in China might rise some skepticism, which we discuss in [the section on China](#China).
- In-ovo sexing.
In-ovo sexing identifies the sex of chicken embryos in fertilized eggs, so that male embryos can be removed before they hatch, instead of being killed as chicks.
As explained in the accompanying article***, it has already reduced the number of male chicks that are killed in some countries, like Germany and France.
We do not account for it, since we found no data on how widely it is used in each country, so our estimates may be too high for those countries.
On a global scale, the difference is small: even if all male chicks in the European Union were spared, the global number of chickens killed would be less than 3% lower.

## Country-specific estimates

FAOSTAT's statistics of animals killed to produce meat (those 88 billion land animals mentioned before) do not include pre-slaughter mortality.
Hence, to be able to compare the impact of meat and eggs under similar assumptions, we do not consider the effect of pre-slaughter mortality in either case.

In this section we compare our default estimate with a more refined one for countries for which there are additional data sources available.

### China
China is the country that contributes the largest amount of eggs and hens to the global total.
With our default estimate, China adds over 5 billion chickens, which is over a third of the total.

Every year, China's National Bureau of Statistics publishes the tonnes of eggs produced from all poultry in its [Statistical Yearbook](https://www.stats.gov.cn/sj/ndsj/2025/html/C12-14.jpg).
For every year from 2015 to 2024, the QCL dataset gives the same figures, to within 0.1%.
For 2007 to 2014, QCL's figures are up to 1.2% lower, because QCL has not taken up the revision that the National Bureau of Statistics made after its 2016 agricultural census.

The National Bureau of Statistics does not publish eggs from hens separately.
QCL's eggs from hens are a fixed share of eggs from all poultry: 85% from 1997 to 2016, and 86% since 2020.

```python
# Eggs from all poultry in China, in million tonnes, from the 2025 Statistical Yearbook of the National Bureau of
# Statistics (table 12-14).
CHINA_NBS_EGGS = {
    2007: 25.467,
    2008: 26.996,
    2009: 27.519,
    2010: 27.769,
    2011: 28.304,
    2012: 28.854,
    2013: 29.055,
    2014: 29.303,
    2015: 30.461,
    2016: 31.605,
    2017: 30.963,
    2018: 31.283,
    2019: 33.090,
    2020: 34.678,
    2021: 34.088,
    2022: 34.564,
    2023: 35.630,
    2024: 35.885,
}

# Eggs from all poultry (item 00001783) and eggs from hens (item 00001062) in China, in tonnes, from QCL.
china_eggs = qcl[
    (qcl["country"] == "China")
    & (qcl["item_code"].astype(str).isin(["00001783", "00001062"]))
    & (qcl["element_code"].astype(str) == "005510")
]
china_eggs = china_eggs.pivot_table(index="year", columns="item_code", values="value", observed=True)

# QCL's eggs from all poultry compared with the official figures.
difference = 100 * (china_eggs["00001783"] / 1e6 / pd.Series(CHINA_NBS_EGGS) - 1)
assert (difference.loc[2015:2024].abs() < 0.1).all(), "QCL's eggs no longer match the official figures for 2015 to 2024."
assert round(difference.loc[2007:2014].abs().max(), 1) == 1.2, "QCL's eggs for 2007 to 2014 have changed."

# QCL's eggs from hens as a share of eggs from all poultry.
hen_share = 100 * china_eggs["00001062"] / china_eggs["00001783"]
assert (hen_share.loc[1997:2016].round() == 85).all(), "QCL's share of eggs from hens in 1997 to 2016 has changed."
assert (hen_share.loc[2020:].round(1) == 86).all(), "QCL's share of eggs from hens since 2020 has changed."
```

For 2024, China's official statistics from the [National Bureau of Statistics](https://www.stats.gov.cn/xxgk/sjfb/tjgb2020/202502/t20250228_1958817.html) report 35.88 million tonnes of eggs from all poultry.
The QCL dataset attributes 30.86 million tonnes of these to hens.
A review of China's egg industry by Sun Congjiao and Yang Ning in the [China Journal of Animal Science](https://fsc.foodmate.net/show.php?itemid=710116) gives a similar figure, 29.41 million tonnes of chicken eggs (and 22.39 million tonnes of high-yield commercial hens).

Hence, we can assume that the tonnes of egg production reported in the QCL dataset are reliable (even though all production values in the QCL dataset are flagged as estimated since 2020).

On the other hand, the yield reported by the QCL dataset is of 184 eggs per hen, and it's also flagged as estimated.
This yield may seem low for a country with a large industrial egg production like China, but it could be plausible if there is also a large share of backyard hens with a low yield.
What's even more worrying is that the yield has been exactly the same since 2017.

It seems therefore plausible that the number of hens (also flagged as estimated) is derived as the result of dividing the production weight by the yield. 
Indeed, from 2017 to 2024, QCL's production of hen eggs in China grew by 16%, while its yield stayed at 184 eggs (9.19 kg) per hen in every year.

```python
# QCL's yield of hen eggs in China since 2017, in eggs and in kg per hen, compared with its growing production.
china_qcl = qcl[
    (qcl["country"] == "China")
    & (qcl["item_code"].astype(str) == "00001062")
    & qcl["element_code"].astype(str).isin(["005413", "005424", "005510"])
]
china_qcl = china_qcl.pivot_table(index="year", columns="element_code", values="value", observed=True).loc[2017:2024]
assert (china_qcl["005413"] == 184).all(), "QCL's eggs per hen in China since 2017 have changed."
assert ((china_qcl["005424"] / 9.19 - 1).abs() < 0.001).all(), "QCL's yield in kg per hen in China since 2017 has changed."
production_growth = china_qcl.loc[2024, "005510"] / china_qcl.loc[2017, "005510"] - 1
assert round(100 * production_growth) == 16, "QCL's production of hen eggs in China has changed."
```

The market research firm Zhuochuang counted on average 1.26 billion commercial hens in production in 2024 ([Xinhua Finance](https://www.cnfin.com/dz-lb/detail/20250122/4178254_1.html)).
The China Animal Agriculture Association reports that a commercial hen lays 17.65 kg of eggs a year ([caaa.cn](https://www.caaa.cn/html/hd/hybg/2023/0316/2722.html)).
Together, these give 22.22 million tonnes of eggs, which is very close to the 22.39 million tonnes that Sun and Yang attribute to high-yield commercial hens.

So high-yield commercial hens can explain well 22.22 million tonnes of eggs, out of the 30.86 million tonnes of eggs reported by the QCL dataset.
That leaves 8.64 million extra tonnes of eggs that could be produced by lower yield hens from small farms and backyard hens.
For simplicity, let's call them backyard hens.
How many of these hens would be needed to explain the remaining 8.64 million tonnes of eggs?

[An article on Kepu China](https://www.kepuchina.cn/more/201712/t20171218_299466.shtml), the science-outreach site of the China Association for Science and Technology, says that most farms and households with free-range hens get laying rates of only 30% to 40%.
This is the percentage of hens that lay an egg on an average day of the year; and hence this translates into 110 to 146 eggs per hen per year.

The QCL hen egg conversion factors assumes 50 g for the typical egg weight in China.
Assuming that weight, backyard hens lay between 5.5 and 7.3 kg of eggs per year.
Therefore, to lay the missing 8.64 million tonnes, China would need between 1.2 and 1.6 billion backyard hens.
Adding up those backyard hens to the 1.26 commercial hens, we get to a total of 2.4 to 2.8 billion hens in China in 2024.
This is somewhat below QCL's 3.36 billion.

But the yield of backyard hens could be significantly lower than 110 to 146 eggs per hen, and the typical hen egg weight could also be lower than 50 g.
Among low-income countries (where egg production is dominated by backyard hens), QCL reports an average yield of 100 eggs per hen, and an average hen egg weight of 45 g.
Assuming those values, the number of hens in China would go up to 3.18 billion, closer to QCL's original value of 3.36 billion.

So this exercise shows that China's estimated number of hens is plausible, but probably on the high end.

But, at the end of the day, what we are trying to estimate is the number of hens killed per year in China, and compare that number with our default estimate.
And here there is an additional parameter that changes significantly between commercial and backyard hens: the laying period.

Using these values, we can estimate the number of hens killed in China in each year from 2020 to 2024, and compare it with our default estimate.
As above, we split China's hens into commercial hens and backyard hens.

For commercial hens:
- The number of commercial hens is Zhuochuang's yearly average.
For 2020 and 2023, which have no yearly average in the reports, we derive it from the next year's average and its change on the year before.
- Their laying period is their age at culling (the average of each year's readings in the table above) minus 18 weeks.
- The number of commercial hens killed per year is the number of commercial hens, times 52 weeks, divided by their laying period.

For backyard hens:
- Their eggs are QCL's hen eggs minus the eggs of commercial hens (17.65 kg per hen per year).
- The number of backyard hens is their eggs divided by their yield.
We take a yield between 4.5 kg (100 eggs of 45 g, as in low-income countries) and 7.3 kg (146 eggs of 50 g, which is a laying rate of 40% with QCL's egg weight for China).
- The same Kepu China article says that backyard hens should be culled after one production cycle of 500 to 600 days, although some farmers keep them for 3 to 4 years or longer.
Minus 18 weeks before they start laying, this gives a laying period of 53 to 68 weeks.
- The number of backyard hens killed per year is the number of backyard hens, times 52 weeks, divided by their laying period.

The lower limit combines the highest yield of backyard hens (7.3 kg) with their longest laying period (68 weeks).
The upper limit combines the lowest yield of backyard hens (4.5 kg) with their shortest laying period (53 weeks).
Backyard hens kept for 3 to 4 years would bring the lower limit further down.
On the other hand, native breeds may start laying later than commercial hens (the Xianju chicken at 160 to 180 days of age, according to [Zhihui Sannong](https://www.pwsannong.com/c/2016-04-13/564176.shtml)), which would shorten their laying period and raise the upper limit.

```python
# Hens killed per year in China: our default estimate, and a lower and an upper limit built from commercial and backyard
# hens, in the years with data on commercial hens.
YEARS = range(2020, 2025)
# Commercial hens in production (yearly average), from Zhuochuang. The 2020 and 2023 values come from the next year's
# average and its change on the year before: the 1.18 billion of 2021 were "down nearly 10%" on 2020 (Sina Finance,
# 17 December 2021), and the 1.259 billion of 2024 were up 5.62% on 2023 (Xinhua Finance, 22 January 2025).
CHINA_COMMERCIAL_HENS = {2020: 1.18e9 / 0.90, 2021: 1.18e9, 2022: 1.170e9, 2023: 1.259e9 / 1.0562, 2024: 1.259e9}
# Age of spent commercial hens at culling, from the industry monitoring data of Zhuochuang Information relayed by Sina
# Finance (see the table above), minus the weeks between hatching and the start of lay. The reports place the start of
# lay four to five months after the chicks are placed; we take 18 weeks, as for the rest of the analysis.
WEEKS_TO_FIRST_LAY = 18
CHINA_CULLING_AGE_DAYS = {
    "2020-05": 456,
    "2021-07": 504,  # Average of 2021 (to mid-December), placed in the middle of the year.
    "2022-11": 532,
    "2022-12": 525,
    "2023-01": 520,
    "2023-03": 524,
    "2023-08": 520,
    "2023-10": 528,
    "2023-12": 524,
    "2024-01": 518,
    "2024-05": 501,
    "2024-09": 532,
    "2025-01": 541,
    "2025-03": 538,
    "2025-04": 536,
    "2025-07": 506,
    "2025-09": 500,
    "2025-12": 480,
}
# Eggs laid by a commercial hen in a year, in kg, from the China Animal Agriculture Association.
COMMERCIAL_YIELD_KG = 17.65
# Eggs laid by a backyard hen in a year, in kg: 100 eggs of 45 g, as in low-income countries, to 146 eggs of 50 g, which
# is a laying rate of 40% (Kepu China) with QCL's egg weight for China.
BACKYARD_YIELD_KG_LOW = 100 * 0.045
BACKYARD_YIELD_KG_HIGH = 146 * 0.050
# Age of backyard hens at culling, in days: one production cycle of 500 to 600 days (Kepu China).
BACKYARD_CULLING_AGE_DAYS_LOW = 500
BACKYARD_CULLING_AGE_DAYS_HIGH = 600

# Hen eggs in China, in tonnes, from QCL.
hen_eggs = qcl[
    (qcl["country"] == "China")
    & (qcl["item_code"].astype(str) == "00001062")
    & (qcl["element_code"].astype(str) == "005510")
]
china_hens = hen_eggs.set_index("year")["value"].reindex(YEARS).to_frame("hen_eggs_tonnes")

# Commercial hens killed: commercial hens, times 52 weeks, divided by their laying period. Their age at culling in each
# year is the average of that year's readings.
culling_age_days = pd.Series(CHINA_CULLING_AGE_DAYS)
culling_age_days = culling_age_days.groupby(culling_age_days.index.str[:4].astype(int)).mean()
china_hens["commercial_hens"] = pd.Series(CHINA_COMMERCIAL_HENS)
china_hens["commercial_hens_killed"] = (
    52 * china_hens["commercial_hens"] / (culling_age_days.reindex(YEARS) / 7 - WEEKS_TO_FIRST_LAY)
)

# Backyard hens: the hen eggs not laid by commercial hens, divided by the yield of a backyard hen.
backyard_eggs_kg = 1000 * china_hens["hen_eggs_tonnes"] - COMMERCIAL_YIELD_KG * china_hens["commercial_hens"]
china_hens["backyard_hens_low"] = backyard_eggs_kg / BACKYARD_YIELD_KG_HIGH
china_hens["backyard_hens_high"] = backyard_eggs_kg / BACKYARD_YIELD_KG_LOW

# Lower limit: the fewest backyard hens, with the longest laying period. Upper limit: the most backyard hens, with the
# shortest laying period.
china_hens["lower_limit"] = china_hens["commercial_hens_killed"] + 52 * china_hens["backyard_hens_low"] / (
    BACKYARD_CULLING_AGE_DAYS_HIGH / 7 - WEEKS_TO_FIRST_LAY
)
china_hens["upper_limit"] = china_hens["commercial_hens_killed"] + 52 * china_hens["backyard_hens_high"] / (
    BACKYARD_CULLING_AGE_DAYS_LOW / 7 - WEEKS_TO_FIRST_LAY
)
china_hens["default_estimate"] = tb_estimates[tb_estimates["country"] == "China"].set_index("year")["hens_killed"]

# The figures for 2024 given in the text.
assert round(backyard_eggs_kg[2024] / 1e9, 2) == 8.64, "The 2024 eggs from backyard hens have changed."
total_hens_2024 = china_hens.loc[2024, ["backyard_hens_low", "backyard_hens_high"]] + china_hens.loc[2024, "commercial_hens"]
assert round(total_hens_2024["backyard_hens_high"] / 1e9, 2) == 3.18, "The 2024 total of hens has changed."
assert round(china_hens.loc[2024, "commercial_hens_killed"] / 1e9, 2) == 1.17, "The 2024 number of commercial hens killed has changed."
assert (
    (china_hens["lower_limit"] < china_hens["default_estimate"]) & (china_hens["default_estimate"] < china_hens["upper_limit"])
).all(), "The default estimate is no longer between the lower and the upper limit."

fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=china_hens.index,
        y=china_hens["lower_limit"] / 1e9,
        name="Lower limit: highest yield and longest laying period of backyard hens",
        line=dict(color="#ff7f0e"),
        hovertemplate="%{y:.2f} billion",
    )
)
fig.add_trace(
    go.Scatter(
        x=china_hens.index,
        y=china_hens["upper_limit"] / 1e9,
        name="Upper limit: lowest yield and shortest laying period of backyard hens",
        line=dict(color="#ff7f0e", dash="dash"),
        fill="tonexty",
        fillcolor="rgba(255, 127, 14, 0.15)",
        hovertemplate="%{y:.2f} billion",
    )
)
fig.add_trace(
    go.Scatter(
        x=china_hens.index,
        y=china_hens["default_estimate"] / 1e9,
        name="Default estimate",
        line=dict(color="#1f77b4"),
        hovertemplate="%{y:.2f} billion",
    )
)
fig.update_yaxes(rangemode="tozero", title_text="Billion hens")
fig.update_xaxes(dtick=1)
fig.update_layout(
    title="Hens killed per year in China",
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="top", y=-0.15),
    height=450,
    margin=dict(t=80),
)
fig.show()
```

[A study by the Chinese Academy of Agricultural Sciences](https://www.zgxmzz.cn/paper/10274) gives 1.233 billion spent laying hens slaughtered in China in 2024.
That count comes from industry monitoring data, so it likely leaves out backyard hens; it is close to our 1.17 billion commercial hens killed in 2024.

### United States
The US Department of Agriculture (USDA) publishes egg and hen data every month.
Three of its indicators are relevant for our analysis:
1. The number of layers sold for slaughter.
These are often called "spent hens": laying hens that are no longer productive enough, and are sold for meat.
2. The number of layers that died or were destroyed.
USDA counts here the layers that were "rendered, died, destroyed, composted, or disappeared for any reason", which includes flocks destroyed in outbreaks of avian influenza.
3. The number of egg-type chicks hatched.
These are chicks of egg-laying breeds (not broiler breeds), both male and female.

Adding up (1) and (2) across all months gives the number of hens that left the laying flock each year, which we take as the number of hens killed.
Adding up (3) across all months, and dividing by 2, gives the number of male chicks killed each year.

USDA's layers also include hens kept for hatching eggs, most of them broiler breeders (62 million of the 377 million layers in 2024).
These belong to meat production rather than egg production.
FAOSTAT's count of laying hens for the United States is the same as USDA's, so our default estimate includes them too.
So, in the following chart, we leave broiler breeders out on both sides:
- Our estimate starts from the laying hens minus the broiler breeders.
- USDA's hens killed exclude the spent broiler breeders slaughtered (reported by USDA as heavy-type mature chickens).

The chart compares our default estimate with USDA's figures of hens killed, male chicks killed, and their sum.
Two things are worth keeping in mind:
- The egg-type chicks hatched are more than the hens that leave the laying flock.
According to USDA, the chicks hatched also include chicks for breeding flocks and for research, and what happens to egg-type chicks before they join the laying flock can vary a lot from year to year.
Male chicks, on the other hand, are all killed.
- USDA's figures include pre-slaughter mortality, which our default estimate does not.
The blue dash-dotted line adds it to our estimate, following Faunalytics (3.5% of pullets die before they start laying, and 8% of hens during the laying period).

```python
# Compare our estimate for the United States with the official figures of the US Department of Agriculture (USDA).
ours = tb_estimates[tb_estimates["country"] == "United States"]
usda = load_table(USDA_TABLE)
assert not ours.empty and not usda.empty, "No data for the United States."
compare = ours.merge(usda, on=["country", "year"])

# USDA's layers include the broiler breeders (hatching egg layers of broiler type), which belong to the meat industry,
# and so does FAOSTAT's count of laying hens, which equals USDA's. Both sides of the comparison leave them out.
# Our estimate: laying hens minus broiler breeders, times the same turnover (52 weeks over the assumed laying period),
# and one male chick per hen.
turnover = compare["hens_killed"] / compare["laying_hens"]
compare["ours_hens_killed"] = (compare["laying_hens"] - compare["average_broiler_type_hatching_egg_layers"]) * turnover
compare["ours_male_chicks_killed"] = compare["ours_hens_killed"]
# USDA: layers sold for slaughter plus layers lost (added up from December of the previous year through November,
# USDA's year for layers), minus the spent broiler breeders slaughtered (reported as heavy-type mature chickens); and
# half of the egg-type chicks hatched, assuming that half of them are male.
compare["usda_hens_killed"] = (
    compare["layers_sold_for_slaughter"] + compare["layers_lost"] - compare["heavy_mature_chickens_slaughtered"]
)
compare["usda_male_chicks_killed"] = compare["egg_type_chicks_hatched"] / 2
for prefix in ["ours", "usda"]:
    compare[f"{prefix}_chickens_killed"] = compare[f"{prefix}_hens_killed"] + compare[f"{prefix}_male_chicks_killed"]
# Our estimate with pre-slaughter mortality, which USDA's figures include: following Faunalytics
# (https://faunalytics.org/global-chick-culling-2019-2024/), 3.5% of pullets die before they start laying, and 8% of
# hens die during the laying period (spread evenly over it, hence the division by 2).
MORTALITY_BEFORE_LAY = 0.035
MORTALITY_IN_LAY = 0.08
compare["ours_chickens_killed_with_mortality"] = compare["ours_hens_killed"] / (1 - MORTALITY_IN_LAY / 2) + compare[
    "ours_male_chicks_killed"
] / ((1 - MORTALITY_BEFORE_LAY) * (1 - MORTALITY_IN_LAY / 2))

# One chart with hens, male chicks and their sum, from our default estimate and from USDA, in millions of birds per year.
# Our hens killed and male chicks killed are the same number, drawn as one line.
assert round((compare["ours_chickens_killed_with_mortality"] / compare["ours_chickens_killed"] - 1).max(), 2) == 0.06, "The mortality correction given in the text has changed."
# Our estimate as published (with broiler breeders) is within about 25% of USDA's figures (median from 2020 to 2024).
recent_us = compare.set_index("year").loc[2020:2024]
error = "Our estimate for the United States is no longer within about 25% of USDA's figures."
assert 0.75 <= (recent_us["usda_hens_killed"] / recent_us["hens_killed"]).median() <= 1.25, error
TRACES = [
    ("ours_chickens_killed", "Default estimate of chickens killed", "#2ca02c", "solid"),
    ("ours_chickens_killed_with_mortality", "Default estimate of chickens killed, adding pre-slaughter mortality", "#1f77b4", "dashdot"),
    ("usda_chickens_killed", "USDA data on chickens killed", "#d62728", "solid"),
    ("ours_hens_killed", "Default estimate of hens killed, same as male chicks killed", "#2ca02c", "dash"),
    ("usda_hens_killed", "USDA data on hens killed", "#d62728", "dash"),
    ("usda_male_chicks_killed", "USDA data on male chicks killed", "#d62728", "dot"),
]
fig = go.Figure()
for column, name, color, dash in TRACES:
    data = compare.dropna(subset=[column])
    fig.add_trace(
        go.Scatter(
            x=data["year"],
            y=data[column] / 1e6,
            name=name,
            line=dict(color=color, dash=dash),
            hovertemplate="%{y:.0f} million",
        )
    )
fig.update_yaxes(rangemode="tozero", title_text="Million birds per year")
fig.update_xaxes(dtick=2)
fig.update_layout(
    title="Chickens killed to produce eggs in the United States",
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="top", y=-0.12),
    height=520,
    margin=dict(t=80),
)
fig.show()
```

The agreement is reasonably good in recent years.
Note, however, that our estimate does not include pre-slaughter mortality, whereas USDA's numbers do.
Adding it would make our estimate about 6% higher.

### European countries
Eurostat publishes the number of chicks of laying hen breeds hatched for use by hatcheries in each European country.
It also publishes the number of these chicks imported and exported.

From these, we get the chicks placed in a country: those hatched, minus those exported, plus those imported.
These are the hens entering the country's laying flock each year.
In a flock of stable size, this equals the number of hens that leave it, and the number of male chicks hatched alongside them.

The chart shows the ten European countries with the most hens killed in our estimate.
It leaves out countries where Eurostat's chicks placed do not measure the hens entering the laying flock:
- Countries without data on chick trade.
- Countries where most chicks hatched are exported, or most chicks placed are imported.
- Italy, which reports more than twice as many cockerels from sexing as laying hen chicks (a category that, according to Eurostat, countries interpret differently).

It also leaves out the United Kingdom, which we compare with Defra's figures in [the section on the United Kingdom](#United-Kingdom).

```python
# Compare our estimate with the laying hen chicks placed in each European country, according to Eurostat.
eurostat = load_table(EUROSTAT_TABLE)
# Chicks placed in a country (hatched, minus exported, plus imported) are the hens entering the laying flock, which in
# a flock of stable size equals the hens leaving it, and the number of male chicks hatched alongside them.
# Countries without data on chick trade have no chicks placed, and are left out here.
compare = eurostat.dropna(subset=["laying_hen_chicks_placed"]).merge(
    tb_estimates[["country", "year", "hens_killed", "laying_hens"]], on=["country", "year"]
)
# The United Kingdom is compared with Defra's figures in its own section.
compare = compare[compare["country"] != "United Kingdom"]
recent = compare[compare["year"].between(2015, 2024)].copy()
recent["ratio"] = recent["laying_hen_chicks_placed"] / recent["hens_killed"]

# Leave out the countries where the chicks placed do not measure the hens entering the laying flock, based on their
# figures from 2015 to 2024.
totals = recent.groupby("country")[
    [
        "laying_hen_chicks_hatched",
        "laying_hen_chicks_exported",
        "laying_hen_chicks_imported",
        "laying_hen_chicks_placed",
        "cockerels_from_sexing_hatched",
    ]
].sum()
# Countries where most chicks hatched are exported, or most chicks placed are imported. Chicks hatched, including those
# exported, are the chicks placed plus exported minus imported (Belgium reports its chicks hatched without the exports).
hatched_with_exports = (
    totals["laying_hen_chicks_placed"] + totals["laying_hen_chicks_exported"] - totals["laying_hen_chicks_imported"]
)
mostly_traded = (totals["laying_hen_chicks_exported"] > 0.5 * hatched_with_exports) | (
    totals["laying_hen_chicks_imported"] > 0.5 * totals["laying_hen_chicks_placed"]
)
# Countries that report more than twice as many cockerels from sexing as laying hen chicks.
many_cockerels = totals["cockerels_from_sexing_hatched"] > 2 * totals["laying_hen_chicks_hatched"]
assert set(totals.index[many_cockerels]) == {"Italy"}, "Countries with many cockerels from sexing are not only Italy."
excluded = set(totals.index[mostly_traded | many_cockerels])

# One panel for each of the ten countries with the most hens killed in our estimate (median from 2015 to 2024).
hens_killed = recent[~recent["country"].isin(excluded)].groupby("country")["hens_killed"].median()
countries = hens_killed.sort_values(ascending=False).index[:10].tolist()
assert len(countries) == 10, "Too few countries with Eurostat data."

# The figures given in the text.
assert {"France", "Spain", "Poland", "Germany", "Romania"} <= set(countries), "Expected countries are not in the chart."
within_20 = (recent["ratio"] - 1).abs() <= 0.2
error = "France, Spain or Poland no longer agree with our estimate within 20% in most years."
assert (within_20.groupby(recent["country"]).mean()[["France", "Spain", "Poland"]] >= 0.8).all(), error
germany = recent[recent["country"] == "Germany"].set_index("year")
assert ((germany.loc[:2020, "ratio"] - 1).abs() <= 0.2).all(), "Germany no longer agrees with our estimate until 2020."
assert (germany.loc[2021:, "ratio"] < 0.8).all(), "Germany's chicks placed are no longer below our estimate from 2021."
error = "Germany's number of hens fell after 2020."
assert (germany.loc[2021:, "laying_hens"] >= germany.loc[2020, "laying_hens"]).all(), error
others = [country for country in countries if country not in ["France", "Spain", "Poland", "Germany"]]
median_ratio = recent.groupby("country")["ratio"].median()
error = "In most of the other countries, the chicks placed are no longer below our estimate."
assert (median_ratio[others] < 1).sum() > len(others) / 2, error
assert median_ratio["Romania"] < 0.05, "Romania's chicks placed are no longer almost zero."

COLUMNS = 5
rows = -(-len(countries) // COLUMNS)
fig = make_subplots(
    rows=rows,
    cols=COLUMNS,
    subplot_titles=countries,
    shared_xaxes="all",
    vertical_spacing=0.12,
    horizontal_spacing=0.04,
)
for index, country in enumerate(countries):
    row, col = divmod(index, COLUMNS)
    ours = tb_estimates[tb_estimates["country"] == country].set_index("year")["hens_killed"]
    placed = compare[compare["country"] == country].set_index("year")["laying_hen_chicks_placed"]
    for series, name, color in [
        (ours, "Our estimate", "#1f77b4"),
        (placed, "Eurostat: laying hen chicks placed", "#d62728"),
    ]:
        fig.add_trace(
            go.Scatter(
                x=series.index,
                y=series / 1e6,
                name=name,
                legendgroup=name,
                showlegend=index == 0,
                mode="lines",
                line=dict(color=color, width=1.5),
                hovertemplate="%{y:.1f} million",
            ),
            row=row + 1,
            col=col + 1,
        )
fig.update_yaxes(rangemode="tozero", tickfont=dict(size=10))
fig.update_xaxes(tickvals=[1990, 2005, 2020], tickangle=0, tickfont=dict(size=10), showticklabels=True)
fig.update_annotations(font_size=13)
fig.update_layout(
    title="Hens killed per year, in million birds: our estimate and the laying hen chicks placed",
    title_font_size=15,
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="bottom", y=1.06, x=0),
    height=230 * rows + 120,
    margin=dict(t=110, l=40, r=20, b=30),
)
fig.show()
```

The two agree within about 20% in most years in France, Spain and Poland.
They also agree in Germany until 2020.
But from 2021, when Germany adopted its ban on chick culling, Germany's chicks placed fell, while its number of hens did not.

In the other countries shown, the chicks placed are mostly below our estimate, and in Romania almost zero.

### United Kingdom
The UK's Department for Environment, Food and Rural Affairs (Defra) counts the breeding and laying fowl on 1 June each year.
This count includes the breeding flock, and pullets that are not yet laying.

Until 2013, FAOSTAT's number of laying hens for the United Kingdom matched Defra's count.
Since 2014, FAOSTAT's figures for the UK are its own estimates, between 3% below and 10% above Defra's count.
So our default estimate for the UK starts from more hens than are actually in lay.

Defra also publishes the number of commercial layer chicks placed by hatcheries each year: the female chicks that enter the laying flock.
In a flock of stable size, this equals the number of hens that leave it each year.
It is also the number of male chicks hatched alongside them.

The chart compares our default estimate with these figures.

### How accurate is the default estimate?
In the United States, the United Kingdom, France, Spain and Poland, which hold most of the hens we can check, our default estimate is within about 25% of the national figures; in several other European countries, it is further above them.
Given the large uncertainties involved, a more sophisticated approach would not be justified.
The most important check is China's, where the default estimate lies within the range that Chinese sources support.

## Laying period

Our calculation relies heavily on the assumption of a 65-week laying period, which means that, in a 52-week year, about 80% of the laying flock is replaced and killed.
[Faunalytics' methodology](https://faunalytics.org/global-chick-culling-2019-2024/) assumes 62 weeks: a traditional single cycle, in which hens start laying at roughly 18 weeks of age and are replaced at roughly 80 weeks.
We adopt 65 weeks as a compromise between the commercial flocks of the United States, where USDA's figures imply 64 to 75 weeks in recent years, and of China, where most industry reports imply 54 to 59 weeks (see the chart below).
As we saw in [the section on the United States](#United-States), this parameter gives an annual number of hens killed 5% to 16% above USDA's figures between 2020 and 2024.

China produces [almost a third of the world's eggs](https://ourworldindata.org/explorers/global-food?tab=line&country=OWID_WRL~CHN&Food=Eggs+from+hens&Metric=Eggs+produced&Per+capita=false), so the laying period of its hens has the largest effect on our global estimate.
Different reports of the market research firm Zhuochuang Information mention that spent hens were culled at [507](https://www.7hcn.com/article/472927-1.html) and [538](https://finance.sina.com.cn/money/future/2025-04-02/doc-ineruann5403612.shtml) days of age.
Since hens start laying at around 18 weeks of age, this means a laying period of 54 to 59 weeks, below our 65 weeks.
[Hendrix Genetics](https://layinghens.hendrix-genetics.com/en/news/chinas-evolving-egg-breeder-market-challenges-and-opportunities/), one of the two main breeders of laying hens, gives a consistent 72 to 80 weeks of age, that is, a laying period of 54 to 62 weeks.
So our adopted 65 weeks is conservative for commercial hens in China.
Adopting a laying period of 54 weeks for China would increase its annual number of killed hens by about 20%, and the global number by about 8%.

On the other hand, in many low- and middle-income countries, mostly in Asia and Africa, egg production is dominated by traditional backyard chickens.
Backyard hens may be kept for longer than 65 weeks, which would mean fewer hens are killed each year, but many may also die from disease before they are killed; we found no data on how long backyard hens are kept.
So a more accurate laying period for those countries may be longer than 65 weeks, in which case our estimated number of hens killed would be too high.
However, FAOSTAT's number of laying hens in most of those countries is flagged as estimated or imputed, so a more sophisticated approach would add complexity without much gain in accuracy.

At a global scale, the contribution of these countries is expected to be small.
A rough way to identify those countries dominated by traditional farms is to select those with a low yield of eggs per bird.
If we select yields below 150 eggs per hen, the resulting set of countries hold about 16% of the world's laying hens in 2024.
If their hens were replaced every 150 weeks instead of 65, our global estimate of chickens killed would fall by less than 10%.

```python
# The figures given above, for 2024, from our default estimate.
WEEKS_IN_LAY = 65
estimates_2024 = tb_estimates[tb_estimates["year"] == 2024].set_index("country")

# Even if all male chicks in the EU were spared by in-ovo sexing, the global number of chickens killed would be less
# than 3% lower.
eu_male_chicks = estimates_2024.loc["European Union (27)", "male_chickens_killed"]
error = "The EU's male chicks are no longer less than 3% of the global number of chickens killed."
assert eu_male_chicks / estimates_2024.loc["World", "chickens_killed"] < 0.03, error

# A laying period of 54 weeks for China, instead of 65 weeks.
china_increase = WEEKS_IN_LAY / 54 - 1
global_increase = china_increase * estimates_2024.loc["China", "hens_killed"] / estimates_2024.loc["World", "hens_killed"]
assert round(100 * china_increase) == 20, "The increase of hens killed in China has changed."
assert round(100 * global_increase) == 8, "The increase of hens killed in the world has changed."

# Countries where FAOSTAT gives fewer than 150 eggs per hen a year, as a rough way to find those dominated by backyard
# hens. Their share of the world's laying hens is computed from FAOSTAT directly, since our estimates start in 1990.
EGGS_PER_HEN_THRESHOLD = 150
# Laying hens (element "Laying") and eggs per hen (element "Yield", in eggs per bird), of item "Hen eggs in shell, fresh".
qcl_hens = qcl[(qcl["item_code"].astype(str) == "00001062") & qcl["element_code"].astype(str).isin(["005313", "005413"])]
# Keep countries only, as in the section on how FAOSTAT counts hens.
area_code = qcl_hens["area_code"].astype(str)
qcl_hens = qcl_hens[area_code.str.isdigit() & (pd.to_numeric(area_code, errors="coerce") < 5000) & (area_code != "351")]
hens = qcl_hens.pivot_table(index=["country", "year"], columns="element_code", values="value", observed=True)
hens = hens.rename(columns={"005313": "laying_hens", "005413": "eggs_per_hen"}).dropna().reset_index()
low_yield_hens = hens["laying_hens"].where(hens["eggs_per_hen"] < EGGS_PER_HEN_THRESHOLD, 0)
share = 100 * low_yield_hens.groupby(hens["year"]).sum() / hens.groupby("year")["laying_hens"].sum()
low_yield_countries = list(hens[(hens["year"] == 2024) & (hens["eggs_per_hen"] < EGGS_PER_HEN_THRESHOLD)]["country"])

# Most of those countries have their laying hens flagged as estimated ("E") or imputed ("I").
laying_flags = qcl_hens[(qcl_hens["year"] == 2024) & (qcl_hens["element_code"].astype(str) == "005313")]
laying_flags = laying_flags[laying_flags["country"].isin(low_yield_countries)]["flag"]
assert laying_flags.isin(["E", "I"]).mean() > 0.5, "Most low-yield countries no longer have estimated or imputed hens."

# Their share of the world's laying hens, and the fall in our global estimate of chickens killed if their hens were
# replaced every 150 weeks instead of 65.
low_yield_killed = estimates_2024.loc[low_yield_countries, "chickens_killed"].sum()
fall = (1 - WEEKS_IN_LAY / 150) * low_yield_killed / estimates_2024.loc["World", "chickens_killed"]
assert round(share[2024]) == 16, "The share of hens in low-yield countries has changed."
assert fall < 0.1, "The fall in our global estimate is no longer below 10%."
```

The chart shows the average time that a hen spends in the laying flock in the United States, according to USDA's figures (see [the section on the United States](#United-States)), and the readings of the age at which spent hens are culled in China, minus 18 weeks of rearing (see [the section on China](#China)), next to the laying period assumed in our default estimate.

```python
# Average time that a hen spends in the laying flock, implied by official figures.
# United States: 52 weeks times the average number of layers for eggs (all layers minus broiler breeders), divided by
# the layers leaving the flock per year (sold for slaughter or lost, minus the spent broiler breeders slaughtered,
# reported as heavy-type mature chickens), from USDA.
usda = load_table(USDA_TABLE)
weeks = usda.dropna(
    subset=[
        "average_layers",
        "average_broiler_type_hatching_egg_layers",
        "layers_sold_for_slaughter",
        "layers_lost",
        "heavy_mature_chickens_slaughtered",
    ]
).copy()
weeks["United States"] = (
    52
    * (weeks["average_layers"] - weeks["average_broiler_type_hatching_egg_layers"])
    / (weeks["layers_sold_for_slaughter"] + weeks["layers_lost"] - weeks["heavy_mature_chickens_slaughtered"])
)

# China: age of spent hens at culling (defined in the China section), minus the weeks between hatching and the start
# of lay.
china = pd.DataFrame({"date": pd.to_datetime(list(CHINA_CULLING_AGE_DAYS)), "days": list(CHINA_CULLING_AGE_DAYS.values())})
china["year"] = china["date"].dt.year + (china["date"].dt.month - 0.5) / 12
china["China"] = china["days"] / 7 - WEEKS_TO_FIRST_LAY

# Laying period assumed in our default estimate, which is the same for all countries and years (with hens).
estimates = tb_estimates[tb_estimates["hens_killed"] > 0]
assumed_weeks = (52 * estimates["laying_hens"] / estimates["hens_killed"]).astype(float).round(1)
assert assumed_weeks.nunique() == 1, "The assumed laying period is not the same for all countries and years."
assumed_weeks = assumed_weeks.iloc[0]

fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=weeks["year"],
        y=weeks["United States"],
        name="United States (USDA)",
        line=dict(color="#d62728"),
        hovertemplate="%{y:.0f} weeks",
    )
)
fig.add_trace(
    go.Scatter(
        x=china["year"],
        y=china["China"],
        name="China (industry reports of the age of spent hens at culling, minus 18 weeks)",
        line=dict(color="#ff7f0e"),
        customdata=china["date"].dt.strftime("%B %Y"),
        hovertemplate="%{customdata}: %{y:.0f} weeks<extra></extra>",
    )
)
fig.add_hline(
    y=assumed_weeks,
    line_dash="dot",
    line_color="#1f77b4",
    annotation_text=f"Laying period assumed in our default estimate: {assumed_weeks:.0f} weeks",
    annotation_position="bottom left",
)
fig.update_yaxes(rangemode="tozero", title_text="Weeks in the laying flock")
fig.update_xaxes(dtick=2)
fig.update_layout(
    title="Average time spent by a hen in the laying flock",
    hovermode="closest",
    legend=dict(orientation="h", yanchor="top", y=-0.15),
    height=450,
    margin=dict(t=80),
)
fig.show()
```
