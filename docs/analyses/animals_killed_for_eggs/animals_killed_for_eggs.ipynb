# Animals killed to produce eggs

[88 billion land animals](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat) were killed in 2024 to produce meat worldwide.
[Almost 90%](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat?stackMode=relative) of those land animals were chicken.

Those are the estimates provided by FAOSTAT, the statistical database of the Food and Agriculture Organization of the United Nations.
But FAOSTAT does not provide an analogous estimate for the number of chickens killed to produce eggs.
And, as we will see later, that number turns out to be surprisingly large.

This technical publication is a companion to the article ***.
Here we explain how to estimate the toll of chickens killed to produce eggs, based on different pieces of information from FAOSTAT and other sources.

## FAOSTAT's crops and livestock products

Every year, FAOSTAT gathers crops and livestock data from each country's official statistics.
FAOSTAT harmonizes the data to ensure figures are consistent with one another and use the same units.
They then update the content of their dataset called [Production: Crops and livestock products](https://www.fao.org/faostat/en/#data/QCL).
From now on, we'll simply call this dataset "QCL", which is the code FAOSTAT uses to identify it.

Unfortunately, not every country reports all of its data every year.
When official data are missing, FAOSTAT may take values from another organization, or estimate them.
To be transparent about the provenance of each data point, they publish each value with a "flag".
This flag clarifies whether a given value is official, taken from another organization, estimated or imputed by FAOSTAT, or missing.

The QCL dataset is indexed by country, year, element (roughly speaking, the metric), and item (roughly, the food product).
For example, the number of chickens killed for meat corresponds to element "Producing Animals/Slaughtered", and item "Meat of chickens, fresh or chilled".
For World in year 2024, this indicator gives 78533923 thousand animals (so, close to 79 billion chickens).
This number is currently flagged as "Estimated value", and it may change in future dataset updates.

To understand the content of the dataset, there are four main sources of information to consider:
- [The QCL methodology document](https://files-faostat.fao.org/production/QCL/QCL_methodology_e.pdf), which is a short document that explains the main concepts.
- [The QCL metadata](https://www.fao.org/faostat/en/#data/QCL/metadata), specifically section 5: Statistical presentation.
- The QCL definitions and standards (which can be accessed from the main QCL data site by clicking on "Definitions and standards"), specifically for "Element", "Item", and "Glossary".
- [The QCL revision note](https://files-faostat.fao.org/production/QCL/Q_Revision_Note_e.pdf), which is a short document explaining a methodological revision carried out in 2016.

The QCL definitions and standards' glossary distinguishes livestock products based on whether they are primary or processed, and whether they come from live or slaughtered animals:

| | From live animals | From slaughtered animals |
|---|---|---|
| **Primary products** | Include milk, eggs, honey, beeswax, fibres of animal origin | Include meat, edible offal, slaughter fats, fresh hides and skins |
| **Processed products** | Include dairy products like butter, cheese, dried eggs | Include bacon, ham, sausages, canned meat, lard, tallow |

Eggs are therefore a primary product of live animals, and chicken meat is a primary product of slaughtered animals.
There are no processed egg or chicken products in QCL.

In the QCL methodology document there are three sections that are relevant for our analysis: livestock numbers, meat, and eggs.

### Livestock numbers

According to the methodology document, they are meant to cover all domestic animals irrespective of their age and the place or purpose of their breeding. However, they point out two caveats:
> In certain countries, data for chickens, ducks and turkeys do not yet seem to represent the total number of these birds.
    
This could mean that some countries count only commercial farms and may undercount their backyard birds.
    
> Certain other countries give a single figure for all poultry; data for these countries are shown under 'Chickens'.
    
This means that for some countries, the number of chickens may include other birds.
This ambiguity may occur even where figures are flagged as official.

The QCL definitions and standards' elements include the following two definitions for livestock numbers:
- Stocks: this element gives the total number of live animals at a given time.
> This variable indicates the number of animals of the species present in the country at the time of enumeration. It includes animals raised either for draft purposes or for meat, eggs and dairy production or kept for breeding. [...] The enumeration to be chosen, when more than one survey is taken, is the closest to the beginning of the calendar year.

- Laying: this element gives the number of birds that lay eggs. However, no specific definition is provided for this element. As we will see later (when speaking about eggs), there is an important ambiguity in how live hens are counted.

### Meat
According to the methodology document, meat is counted where the animal is slaughtered, not where it was born or raised. They mention some inconsistencies in reporting (e.g. the US includes giblets, and some countries may or may not include them). They also explain that:
> All data shown relate to total meat production, that is, from both commercial and farm slaughter. [...]

This means that the number of chickens killed for meat production includes both chickens from commercial slaughterhouses and slaughtered on farms. However, the QCL metadata softens this claim, saying that:
> "Slaughtering" is measured through activity of slaughterhouses, e.g. production of marketable meat for human consumption. Estimates of ‘other slaughtering’ can be added for a more accurate picture of meat production. 

This means that the number of chickens killed for meat comes from slaughterhouse records. Animals killed elsewhere, on farms or in backyards, are either missing, or may be added as an estimate.

### Eggs

The QCL dataset includes the following elements per country and year:
- Tonnes of egg produced: element "Production" (in tonnes).
- Number of eggs produced (element "Production", in thousands).
- Number of live chickens (element "Stocks", in thousands; this element refers to the item "Chickens", which includes all live chickens, not only laying hens).
- Number of laying hens (element "Laying", in thousands).
- Number of eggs produced per hen (element "Yield", in number per animal).

And the QCL dataset includes the following items:
- Eggs from hens: item "Hen eggs in shell, fresh".
- Eggs from other birds: item "Eggs from other birds in shell, fresh, n.e.c.".
- Eggs from all birds: item "Eggs Primary".

But some countries report only one or a few of those elements, and FAOSTAT needs to derive one from the other.
To do that, they assume an average egg weight (how many grams a typical egg weighs) and/or an average lay rate (how many eggs a typical hen lays per year).
These values are taken from official statistics where possible; otherwise FAOSTAT assumes them.
This is explained in the methodology document:
> Some countries have no statistics on egg production, and estimates had to be derived from such related data as chicken or total poultry numbers and reported or assumed rates of egg laying. Most of the countries that have statistics on egg production report either the total weight of eggs or the numbers of eggs produced; data on numbers have been converted into weight, using official conversion factors wherever possible.

QCL includes eggs from all types of hens, and it includes eggs for hatching (fertilized eggs that are not eaten, but hatched into chicks for both egg-laying and meat flocks), but it excludes wasted eggs that did not make it out of the farms (e.g. eggs that broke in the farm).
And they include eggs from both traditional farms (small flocks of backyard hens) and modern ones (large-scale, intensive commercial poultry farms).
This is explained in the QCL definitions and standards' item for "Eggs Primary":
> [...] Egg production by type of poultry should refer to the total production of eggs in the shell by all types of hens in both the traditional sector (individually owned small flocks) and the modern sector (large-scale, intensive commercial poultry farms). Total productions include eggs for hatching but exclude waste of farms. Countries should report in terms of both numbers and weight.


## Annex: additional caveats and assumptions

### Does FAOSTAT measure hens at a point in time, or the cumulative number of hens?

The QCL definitions and standards' glossary includes a very important explanation for the term "Egg production":
> Covers all domestic birds which have contributed to egg production during the year, wherever they lay and the corresponding total production, including eggs intended to be used for hatching but excluding waste on farms.

This sentence has two possible interpretations, with crucially different consequences.
"All domestic birds which have contributed to egg production during the year" can mean:
1. All individual birds that laid eggs during the year: this means that the "Laying" element gives the **cumulative** number of birds that have laid eggs in the year.
2. All kinds of domestic birds that lay eggs: in this case, the sentence says nothing about how birds are counted, and the "Laying" element would give the number of birds that lay eggs **at any given time**.

Note that each interpretation would lead to very different numbers:
Imagine a farm that produces 30,000 eggs a year with 100 hens in lay, and replaces each hen after one year, spreading the replacements evenly over the year.
Under Option 2, the "Laying" element should be 100 (with a yield of 300 eggs per hen).
Under Option 1, "Laying" should report 200 hens (with a yield of 150 eggs per hen): the 100 that were there at the beginning of the year, plus the 100 that replaced them during the year.

Based purely on semantics, Option 1 seems to be the more natural interpretation of FAOSTAT's explanation.
All domestic birds which have contributed to egg production is the cumulative number of hens that were in lay during the year.

However, Option 2 could be interpreted as a clarification that the data includes all species that lay eggs, e.g. hens, ducks, and pigeons, and not just hens.
Option 2 could also be interpreted as a clarification that the data includes breeder hens (of chickens raised both for eggs and chickens raised for meat), and not just laying hens.

Regardless of semantic, the most important question is: which interpretation is more likely the one FAOSTAT assumed?
When asking different experts on eggs and hens data, this was raised as an open question.
To me, however, there are strong reasons to believe that Option 2 is more plausible.

Firstly, the QCL revision note clarifies that data on agricultural productivity (e.g. yield) is not collected.
FAOSTAT derives it by dividing outputs by inputs; in the case of yield per hen, it is estimated as number of eggs divided by number of hens.
Indeed, one can easily demonstrate that FAOSTAT's yield is numerically consistent with the division of egg production (the "Production" element) by the number of laying hens (the "Laying" element).

The calculated yield for most countries with a heavy industrial egg production is between roughly 280 and 320 eggs per bird, which is very plausible.

If FAOSTAT expected the number of laying hens to be cumulative, this calculation would not be a meaningful representation of productivity; the derived yields would also be significantly lower than that.
The fact that FAOSTAT does this calculation is evidence that they assume Option 2 above.
Hence, the element "Laying" gives the number of laying birds at a point in time (and not the cumulative number of birds during the year).

Secondly, even though the QCL's definitions and standards do not provide an explicit explanation of the "Laying" element, they do provide one for the "Stocks" element.
Stocks are very clearly defined as the number of animals in the country at the time of enumeration.

Finally, reporting the number of animals at a point in time seems to be more feasible in practice than reporting a cumulative over the year.
