# Animals killed to produce eggs

[88 billion land animals](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat) were killed in 2024 to produce meat worldwide.
[Almost 90%](https://ourworldindata.org/grapher/land-animals-slaughtered-for-meat?stackMode=relative) of those land animals were chicken.

Those are the estimates provided by FAOSTAT, the statistical database of the Food and Agriculture Organization of the United Nations.
But FAOSTAT does not provide an analogous estimate for the number of chickens killed to produce eggs.
And, as we will see later, that number is estimated to be surprisingly large.

This technical publication is a companion to the article ***.
Here we explain how to estimate the toll of chickens killed to produce eggs, based on different pieces of information from FAOSTAT and other sources.

## FAOSTAT's crops and livestock products

Every year, FAOSTAT gathers crops and livestock data from each country's official statistics.
FAOSTAT harmonizes the data to ensure figures are consistent with one another and use the same units.
They then update the content of their dataset called [Production: Crops and livestock products](https://www.fao.org/faostat/en/#data/QCL).
From now on, we'll simply call this dataset "QCL", which is the code FAOSTAT uses to identify it.

Unfortunately, not every country reports all of its data every year.
When official data are missing, FAOSTAT may take values from another organization, or estimate them.
To be transparent about the provenance of each data point, they publish each value with a "flag".
This flag clarifies whether a given value is official, taken from another organization, estimated or imputed by FAOSTAT, or missing.

The QCL dataset is indexed by country, year, element (roughly speaking, the metric), and item (roughly, the food product).
For example, the number of chickens killed for meat corresponds to element "Producing Animals/Slaughtered", and item "Meat of chickens, fresh or chilled".
For World in year 2024, this indicator gives 78533923 thousand animals (so, close to 79 billion chickens).
This number is currently flagged as "Estimated value", and it may change in future dataset updates.

To understand the content of the dataset, there are four main sources of information to consider:
- [The QCL methodology](https://files-faostat.fao.org/production/QCL/QCL_methodology_e.pdf), which is a short document that explains the main concepts.
- [The QCL metadata](https://www.fao.org/faostat/en/#data/QCL/metadata), specifically section 5: Statistical presentation.
- [The QCL definitions and standards](https://www.fao.org/faostat/en/#data/QCL) (which can be accessed from the main QCL data site by clicking on "Definitions and standards"), specifically for "Element", "Item", and "Glossary".
- [The QCL revision note](https://files-faostat.fao.org/production/QCL/Q_Revision_Note_e.pdf), which is a short document explaining a methodological revision carried out in 2016.

The QCL definitions and standards' glossary distinguishes livestock products based on whether they are primary or processed, and whether they come from live or slaughtered animals:

| | From live animals | From slaughtered animals |
|---|---|---|
| **Primary products** | Include milk, eggs, honey, beeswax, fibres of animal origin | Include meat, edible offal, slaughter fats, fresh hides and skins |
| **Processed products** | Include dairy products like butter, cheese, dried eggs | Include bacon, ham, sausages, canned meat, lard, tallow |

Eggs are therefore a primary product of live animals, and chicken meat is a primary product of slaughtered animals.
There are no processed egg or chicken products in QCL.

In the QCL methodology there are three sections that are relevant for our analysis: livestock numbers, meat, and eggs.

### Livestock numbers

According to the methodology document, their livestock numbers are meant to cover all domestic animals irrespective of their age and the place or purpose of their breeding. However, they point out two caveats:
> In certain countries, data for chickens, ducks and turkeys do not yet seem to represent the total number of these birds.
    
This could mean that some countries count only commercial farms and may undercount their backyard birds.
    
> Certain other countries give a single figure for all poultry; data for these countries are shown under 'Chickens'.
    
This means that for some countries, the number of chickens may include other birds.
This ambiguity may occur even where figures are flagged as official.

The QCL definitions and standards' elements include the following two definitions for livestock numbers:
- Stocks: this element gives the total number of live animals at a given time.
> This variable indicates the number of animals of the species present in the country at the time of enumeration. It includes animals raised either for draft purposes or for meat, eggs and dairy production or kept for breeding. [...] The enumeration to be chosen, when more than one survey is taken, is the closest to the beginning of the calendar year.

- Laying: this element gives the number of birds that lay eggs. However, no specific definition is provided for this element. As we explain in Appendix ***, there is an important ambiguity in how live hens are counted.

### Meat
According to the methodology document, meat is counted where the animal is slaughtered, not where it was born or raised. They mention some inconsistencies in reporting (e.g. the US includes giblets, and some countries may or may not include them). They also explain that:
> All data shown relate to total meat production, that is, from both commercial and farm slaughter. [...]

This means that the number of chickens killed for meat production includes both chickens from commercial slaughterhouses and those killed on farms. However, the QCL metadata softens this claim, saying that:
> "Slaughtering" is measured through activity of slaughterhouses, e.g. production of marketable meat for human consumption. Estimates of ‘other slaughtering’ can be added for a more accurate picture of meat production. 

This means that the number of chickens killed for meat comes primarily from slaughterhouse records. Animals killed elsewhere, on farms or in backyards "can be added", which implies that they may often be missing, or may be included as an estimate.

### Eggs

The QCL dataset includes the following elements per country and year:
- Tonnes of egg produced (element "Production", in tonnes).
- Number of eggs produced (element "Production", in thousands).
- Number of live animals (element "Stocks", in thousands).
- Number of laying animals (element "Laying", in thousands).
- Number of eggs produced per hen (element "Yield", in number per animal).

And the QCL dataset includes the following items:
- Eggs from hens (item "Hen eggs in shell, fresh").
- Eggs from other birds (item "Eggs from other birds in shell, fresh, n.e.c.").
- Eggs from all birds (item "Eggs Primary").

Some countries report only one or a few of those elements, and FAOSTAT needs to derive one based on the other.
To do that, FAOSTAT may assume an average egg weight (how many grams a typical egg weighs) and/or an average lay rate (how many eggs a typical hen lays per year).
These values are taken from official statistics where possible; otherwise FAOSTAT assumes them.
This is explained in the QCL methodology:
> Some countries have no statistics on egg production, and estimates had to be derived from such related data as chicken or total poultry numbers and reported or assumed rates of egg laying. Most of the countries that have statistics on egg production report either the total weight of eggs or the numbers of eggs produced; data on numbers have been converted into weight, using official conversion factors wherever possible.

QCL includes eggs from all types of hens, and it includes eggs for hatching (fertilized eggs that are not eaten, but hatched into chicks for both egg-laying and meat flocks), but it excludes wasted eggs that did not make it out of the farms (e.g. eggs that broke in the farm).
They include eggs from both traditional farms (small flocks of backyard hens) and modern ones (large-scale, intensive commercial poultry farms).
This is explained in the QCL definitions and standards' item for "Eggs Primary":
> [...] Egg production by type of poultry should refer to the total production of eggs in the shell by all types of hens in both the traditional sector (individually owned small flocks) and the modern sector (large-scale, intensive commercial poultry farms). Total productions include eggs for hatching but exclude waste of farms. Countries should report in terms of both numbers and weight.

Finally, the QCL definitions and standards' glossary includes a very important explanation for the term "Egg production":
> Covers all domestic birds which have contributed to egg production during the year, wherever they lay and the corresponding total production, including eggs intended to be used for hatching but excluding waste on farms.

As we explain in Appendix ***, this explanation carries some ambiguity.
For reasons we justify in that appendix, we assume that the number of domestic birds counted for egg production is the count of all laying birds at a given point in time (and not the total number of birds that have laid during the year).

```python
import contextlib
import io

import pandas as pd
import plotly.graph_objects as go
from owid.catalog import fetch
from plotly.subplots import make_subplots

# Tables from our catalog that are used in this notebook.
# Our estimates of the number of chickens killed to produce eggs.
CHICKENS_KILLED_FOR_EGGS_TABLE = "garden/animal_welfare/2026-09-14/chickens_killed_for_eggs/chickens_killed_for_eggs"
# USDA's figures of laying hens and egg-type chicks in the United States.
USDA_TABLE = "garden/usda_nass/2026-10-18/us_laying_hens_and_chicks/us_laying_hens_and_chicks"
# Defra's figures of laying hens and chicks in the United Kingdom.
DEFRA_TABLE = "garden/defra/2026-10-18/uk_laying_hens_and_chicks/uk_laying_hens_and_chicks"
# Eurostat's figures of laying hen chicks hatched, imported and exported in European countries.
EUROSTAT_TABLE = "garden/eurostat/2026-10-18/laying_hen_chicks/laying_hen_chicks"


def load_table(path):
    """Load a table from our public catalog or, if it is not published there yet, from a local ETL checkout.

    The table is returned as a plain dataframe: this notebook only plots the data, and does not need its metadata.
    """
    try:
        # NOTE: fetch() prints a progress spinner, which is silenced here.
        with contextlib.redirect_stdout(io.StringIO()):
            return pd.DataFrame(fetch(path).reset_index())
    except Exception:
        from owid.catalog import Dataset

        from etl.paths import DATA_DIR

        dataset_path, table_name = path.rsplit("/", 1)
        return pd.DataFrame(Dataset(DATA_DIR / dataset_path).read(table_name))


# Load our estimates for all countries and years.
tb_estimates = load_table(CHICKENS_KILLED_FOR_EGGS_TABLE)
```

## Simple estimate
The main goal of this document is to estimate the number of chickens that are killed to produce eggs.
In this section we will carry out that estimate following a very simple approach.

We start by estimating the number of laying hens killed every year.
For that, we use FAOSTAT's data on the number of live hens (element "Laying", item "Hen eggs in shell, fresh"): this indicator gives the number of hens alive at a given time.
In 2024, the global number of hens alive was around 9 billion.

A young female chicken (a pullet) in a typical commercial system takes around 18 weeks (the rearing period) from the moment of hatching until she starts laying eggs.
Between 18 and about 80 weeks of age (the laying period), the hen lays eggs regularly.
After this period the hen is killed, as she is no longer considered productive enough.
Both the rearing and the laying period vary with the breed and the production system, and both are usually longer in traditional systems.
We will assume a laying period of 62 weeks, as explained in Appendix ***.

Given that there are around 9 billion hens alive at any given time, and each of them lasts for about 62 weeks before it's slaughtered and replaced by another one, then

9 billion hens / laying period x ( 1 laying period / 62 week ) x ( 52 weeks / 1 year ) ~ 7.5 billion hens / year

In other words, roughly 7.5 billion hens were slaughtered worldwide in 2024.

For each female chick hatched, on average one male chick is hatched too.
This relies on the assumption that chicks are equally likely to hatch male or female, which is supported by basic biology.
As we explain in the accompanying article***, the vast majority of those male chicks are killed soon after hatching.
If we assume, to begin with, that all those chicks are culled, this leads to 7.5 billion male chicks killed in 2024.

Therefore, the global number of chickens brought into existence and killed to produce eggs is around 15 billion in 2024.

This simple approach to estimate the number of chickens killed per year can also be performed at the country level.
When we do that, we see that the countries killing the largest number of chickens in 2024 were:
- China, at 5.6 billion,
- India, at 1.1 billion,
- Indonesia, at 0.70 billion,
- United States, at 0.63 billion, and
- Brazil, at 0.47 billion.  

### Limitations on the simple estimate
Some factors push the previous estimate up, and others pull it down.

#### Arguments that push the estimate up
- Pre-slaughter mortality.
A fraction of pullets die before they become laying hens, and a fraction of laying hens die, or are killed, before they reach the slaughter age.
Once we correct our estimates to include these pre-slaughter deaths (see Appendix ***), the number goes up to about 8 billion hens, leading to a total of 16 billion chickens killed in 2024.
This figure is consistent with [a recent publication by Faunalytics](https://faunalytics.org/global-chick-culling-2019-2024/), that puts the number of male chicks culled at 8 billion per year.
- Disease outbreaks.
The estimate leaves out hens culled in disease outbreaks such as avian influenza.
- Inconsistent data for Indonesia.
FAOSTAT's number of hens in Indonesia (the third largest egg producer worldwide) is too small to account for the eggs reported, which suggests that a significant fraction of hens could be missing from the count.
- Shorter laying period.
The typical laying period among China's commercial hens could be shorter than our adopted 62 weeks, which would increase the global number of chickens killed.

#### Arguments that push the estimate down
- Backyard hens.
As we discuss in Appendix ***, most of Africa and part of Asia is dominated by backyard hens, with laying periods significantly longer than our assumed 62 weeks. Our estimates are therefore likely overestimating the number of chickens killed in those countries; however, on a global scale, their contribution is small.
- Pullets.
For some countries, FAOSTAT's number of laying hens includes pullets. For those countries, our simple estimate would need to account for the additional 18 weeks of life of the hens before they start laying (so, we would need to replace our 62 by 80 weeks). This issue is known for the UK, but it can also affect other countries.
- Breeders.
FAOSTAT's number of laying hens may include breeders for some countries.
They are also part of the production system, so it is reasonable to include them in the calculation, just like any other type of hen.
But among those hens there are also broiler breeders: hens whose fertilized eggs will become chickens used for meat.
The number of broiler breeders killed should therefore be attributed to meat production, not to egg production.
We expect breeders to typically account for less than 10% of all hens alive, most of them broiler breeders.
- In-ovo sexing.
As explained in the accompanying article***, in-ovo sexing has already reduced the number of male chicks that are killed in Germany and France.
On a global scale, the difference is small, but it is an important correction at the country level.
- Inconsistent data for China.
The number of hens in China reported by FAOSTAT is likely to be overestimated.
This correction, as we will see in Section ***, makes a significant difference in both China's total figures and the global ones.

## Refined estimate

FAOSTAT's statistics of animals killed to produce meat (those 88 billion land animals mentioned before) do not include pre-slaughter mortality.
Hence, to be able to compare the impact of meat and eggs under similar assumptions, we do not consider the effect of pre-slaughter mortality in either case.

The number of hens and male chicks killed per year can also be estimated from official sources for several countries.
We will now attempt to compare our estimate with those from different sources for specific countries.
Some countries (like the UK and EU members) will be inspected simply because we already have access to reliable data, even though they are not among the most relevant in terms of egg production.
As we'll see later, China is the most relevant country in terms of global egg production, and for which access to reliable data is limited.

### United States
The US Department of Agriculture (USDA) publishes egg and hen data on a monthly basis.
Three indicators are relevant for our analysis:
1. The number of layers sold for slaughter. These are often called "spent hens": laying hens that are no longer productive enough, and are sold for meat.
2. The number of layers that died or were destroyed. USDA counts here the layers that were "rendered, died, destroyed, composted, or disappeared for any reason", which includes flocks destroyed in outbreaks of avian influenza.
3. The number of egg-type chicks hatched. These are chicks of egg-laying breeds (not broiler breeds), both male and female.

Adding up (1) and (2) across all months gives the number of hens that left the laying flock each year, which we take as the number of hens killed.
Adding up (3) across all months, and dividing by 2, gives the number of male chicks killed each year.
USDA's layers include hens kept for hatching eggs, most of them broiler breeders (62 million of the 377 million layers in 2024), which belong to meat production rather than egg production.
FAOSTAT's count of laying hens for the United States is the same as USDA's, so our simple estimate includes them too.
The following chart compares our simple estimate with USDA's figures of hens killed, male chicks killed, and their sum, leaving broiler breeders out on both sides: our estimate starts from the laying hens minus the broiler breeders, and USDA's hens killed exclude the spent broiler breeders slaughtered (reported by USDA as heavy-type mature chickens).
Note that the egg-type chicks hatched are more than the hens that leave the laying flock: USDA's layer survey covers producers with at least 30,000 layers, and, as USDA notes, the disposition of egg-type chicks before placement into the laying flock varies significantly (some go to smaller flocks or abroad, or die before laying), whereas all male chicks hatched are killed.
USDA's figures include pre-slaughter mortality, which our simple estimate does not; the blue dash-dotted line adds it to our estimate, following Faunalytics (3.5% of pullets die before they start laying, and 8% of hens during the laying period).

```python
# Compare our estimate for the United States with the official figures of the US Department of Agriculture (USDA).
ours = tb_estimates[tb_estimates["country"] == "United States"]
usda = load_table(USDA_TABLE)
assert not ours.empty and not usda.empty, "No data for the United States."
compare = ours.merge(usda, on=["country", "year"])

# USDA's layers include the broiler breeders (hatching egg layers of broiler type), which belong to the meat industry,
# and so does FAOSTAT's count of laying hens, which equals USDA's. Both sides of the comparison leave them out.
# Our estimate: laying hens minus broiler breeders, times the same turnover (52 weeks over the assumed laying period),
# and the same share of male chicks spared by in-ovo sexing.
turnover = compare["hens_killed"] / compare["laying_hens"]
male_chicks_per_hen = compare["male_chickens_killed"] / compare["hens_killed"]
compare["ours_hens_killed"] = (compare["laying_hens"] - compare["average_broiler_type_hatching_egg_layers"]) * turnover
compare["ours_male_chicks_killed"] = compare["ours_hens_killed"] * male_chicks_per_hen
# USDA: layers sold for slaughter plus layers lost (added up from December of the previous year through November,
# USDA's year for layers), minus the spent broiler breeders slaughtered (reported as heavy-type mature chickens); and
# half of the egg-type chicks hatched, assuming that half of them are male.
compare["usda_hens_killed"] = (
    compare["layers_sold_for_slaughter"] + compare["layers_lost"] - compare["heavy_mature_chickens_slaughtered"]
)
compare["usda_male_chicks_killed"] = compare["egg_type_chicks_hatched"] / 2
for prefix in ["ours", "usda"]:
    compare[f"{prefix}_chickens_killed"] = compare[f"{prefix}_hens_killed"] + compare[f"{prefix}_male_chicks_killed"]
# Our estimate with pre-slaughter mortality, which USDA's figures include: following Faunalytics
# (https://faunalytics.org/global-chick-culling-2019-2024/), 3.5% of pullets die before they start laying, and 8% of
# hens die during the laying period (spread evenly over it, hence the division by 2).
MORTALITY_BEFORE_LAY = 0.035
MORTALITY_IN_LAY = 0.08
compare["ours_chickens_killed_with_mortality"] = compare["ours_hens_killed"] / (1 - MORTALITY_IN_LAY / 2) + compare[
    "ours_male_chicks_killed"
] / ((1 - MORTALITY_BEFORE_LAY) * (1 - MORTALITY_IN_LAY / 2))

# One chart with hens, male chicks and their sum, from our simple estimate and from USDA, in millions of birds per year.
# Our hens killed and male chicks killed are the same number (there is no in-ovo sexing in the US), drawn as one line.
assert (compare["ours_hens_killed"] - compare["ours_male_chicks_killed"]).abs().max() < 1, "Expected equal lines."
TRACES = [
    ("ours_chickens_killed", "Simple estimate of chickens killed", "#2ca02c", "solid"),
    ("ours_chickens_killed_with_mortality", "Simple estimate of chickens killed, adding pre-slaughter mortality", "#1f77b4", "dashdot"),
    ("usda_chickens_killed", "USDA data on chickens killed", "#d62728", "solid"),
    ("ours_hens_killed", "Simple estimate of hens killed, same as male chicks killed", "#2ca02c", "dash"),
    ("usda_hens_killed", "USDA data on hens killed", "#d62728", "dash"),
    ("usda_male_chicks_killed", "USDA data on male chicks killed", "#d62728", "dot"),
]
fig = go.Figure()
for column, name, color, dash in TRACES:
    data = compare.dropna(subset=[column])
    fig.add_trace(
        go.Scatter(
            x=data["year"],
            y=data[column] / 1e6,
            name=name,
            line=dict(color=color, dash=dash),
            hovertemplate="%{y:.0f} million",
        )
    )
fig.update_yaxes(rangemode="tozero", title_text="Million birds per year")
fig.update_xaxes(dtick=2)
fig.update_layout(
    title="Chickens killed to produce eggs in the United States",
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="top", y=-0.12),
    height=520,
    margin=dict(t=80),
)
fig.show()
```

The agreement is reasonably good in recent years.
Once thing to notice, however, is that our estimate does not include pre-slaughter mortality, whereas USDA's numbers do.
If this correction was added to our estimate, it would be about 10% higher.

### United Kingdom
FAOSTAT's number of laying hens for the United Kingdom matches Defra's count of breeding and laying fowl at 1 June, which includes pullets that are not yet laying and the breeding flock.
Our simple estimate for the UK therefore starts from more hens than are actually in lay.
Defra also publishes the number of commercial layer chicks placed by hatcheries each year: the female chicks that enter the laying flock.
In a flock of stable size, this equals the number of hens that leave it each year, and it is also the number of male chicks hatched alongside them.
The chart compares our simple estimate with these figures.

```python
# Compare our estimate for the United Kingdom with the official figures of Defra.
ours = tb_estimates[tb_estimates["country"] == "United Kingdom"]
defra = load_table(DEFRA_TABLE)
assert not ours.empty and not defra.empty, "No data for the United Kingdom."

# Show the years with Defra data on chicks placed.
first_year = defra.dropna(subset=["layer_chicks_placed"])["year"].min()
ours = ours[ours["year"] >= first_year]
defra = defra[defra["year"] >= first_year]

# Left: the number of laying hens we start from (FAOSTAT) against Defra's counts of birds at 1 June.
# Right: our estimate of hens killed per year against the female layer chicks placed by hatcheries, which in a flock of
# stable size equals the hens leaving it (and the number of male chicks hatched alongside them).
fig = make_subplots(rows=1, cols=2, subplot_titles=["Laying hens", "Hens killed per year"])
for name, data, column, color, dash in [
    ("FAOSTAT: laying hens (our starting point)", ours, "laying_hens", "#1f77b4", "solid"),
    ("Defra: breeding and laying fowl at 1 June", defra, "total_breeding_and_laying_fowl", "#d62728", "solid"),
    ("Defra: hens and pullets laying eggs for eating at 1 June", defra, "hens_and_pullets_laying_eggs_for_eating", "#d62728", "dot"),
]:
    fig.add_trace(
        go.Scatter(x=data["year"], y=data[column] / 1e6, name=name, line=dict(color=color, dash=dash), hovertemplate="%{y:.1f} million"),
        row=1,
        col=1,
    )
for name, data, column, color in [
    ("Our estimate", ours, "hens_killed", "#1f77b4"),
    ("Defra: commercial layer chicks placed", defra, "layer_chicks_placed", "#d62728"),
]:
    fig.add_trace(
        go.Scatter(x=data["year"], y=data[column] / 1e6, name=name, line=dict(color=color), hovertemplate="%{y:.1f} million"),
        row=1,
        col=2,
    )
fig.update_yaxes(rangemode="tozero", title_text="Million birds", row=1, col=1)
fig.update_yaxes(rangemode="tozero", row=1, col=2)
fig.update_xaxes(dtick=2)
fig.update_layout(
    title="Laying hens and hens killed in the United Kingdom",
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="top", y=-0.15),
    height=480,
    margin=dict(t=80),
)
fig.show()
```

### European Union members
Eurostat publishes the number of chicks of laying hen breeds hatched for use by hatcheries in each European country, and the chicks imported and exported.
The chicks placed in a country (hatched, minus exported, plus imported) are the hens entering its laying flock each year, which in a flock of stable size equals the hens that leave it, and the number of male chicks hatched alongside them.
The chart compares our simple estimate with the chicks placed in every country with data, ordered by the number of chicks placed.
Where the two differ widely, the cause is in the inputs: FAOSTAT's count includes pullets in the Netherlands and the United Kingdom, Germany's hatcheries halved their output after the ban on chick culling in 2022, and Italy's hatcheries reclassified their laying chicks in 2015.

```python
# Compare our estimate with the laying hen chicks placed in each European country, according to Eurostat.
eurostat = load_table(EUROSTAT_TABLE)
# Chicks placed in a country (hatched, minus exported, plus imported) are the hens entering the laying flock, which in
# a flock of stable size equals the hens leaving it, and the number of male chicks hatched alongside them.
compare = eurostat.dropna(subset=["laying_hen_chicks_placed"]).merge(
    tb_estimates[["country", "year", "hens_killed"]], on=["country", "year"]
)
# One panel per country, ordered by the number of chicks placed (largest first).
countries = compare.groupby("country")["laying_hen_chicks_placed"].median().sort_values(ascending=False).index.tolist()
assert len(countries) >= 10, "Too few countries with Eurostat data."

COLUMNS = 4
rows = -(-len(countries) // COLUMNS)
fig = make_subplots(rows=rows, cols=COLUMNS, subplot_titles=countries, vertical_spacing=0.05, horizontal_spacing=0.05)
for index, country in enumerate(countries):
    row, col = divmod(index, COLUMNS)
    data = compare[compare["country"] == country]
    for column, name, color in [
        ("hens_killed", "Our estimate", "#1f77b4"),
        ("laying_hen_chicks_placed", "Eurostat: laying hen chicks placed", "#d62728"),
    ]:
        fig.add_trace(
            go.Scatter(
                x=data["year"],
                y=data[column] / 1e6,
                name=name,
                legendgroup=name,
                showlegend=index == 0,
                line=dict(color=color),
                hovertemplate="%{y:.1f} million",
            ),
            row=row + 1,
            col=col + 1,
        )
fig.update_yaxes(rangemode="tozero")
fig.update_xaxes(tickformat="d")
fig.update_layout(
    title="Hens killed per year, in million birds: our estimate and the laying hen chicks placed",
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="top", y=-0.03),
    height=240 * rows,
    margin=dict(t=80),
)
fig.show()
```

## Appendix: additional caveats and assumptions

### Main assumption: laying period

Our calculation relies heavily on the assumption of a 62-week laying period, which means that, in a 52-week year, about 84% of the laying flock is replaced and killed.
We adopt this parameter following [Faunalytics' methodology](https://faunalytics.org/global-chick-culling-2019-2024/), which describes it as a traditional single cycle, in which hens start laying at roughly 18 weeks of age and are replaced at roughly 80 weeks, so they spend about 62 weeks in lay.
As we saw in Section ***, this parameter leads to an annual number of hens killed that is in very good agreement with the figures reported by the USDA.

China produces [almost a third of the world's eggs](https://ourworldindata.org/explorers/global-food?tab=line&country=OWID_WRL~CHN&Food=Eggs+from+hens&Metric=Eggs+produced&Per+capita=false), so the laying period of its hens has the largest effect on our global estimate.
Different reports of Zhuochuang Information, the main provider of poultry market data in China, mention that spent hens were culled at [507](https://www.7hcn.com/article/472927-1.html) and [538](https://finance.sina.com.cn/money/future/2025-04-02/doc-ineruann5403612.shtml) days of age.
Since hens start laying at around 18 weeks of age, this means a laying period of 54 to 59 weeks, slightly below our 62 weeks.
[Hendrix Genetics](https://layinghens.hendrix-genetics.com/en/news/chinas-evolving-egg-breeder-market-challenges-and-opportunities/), one of the two main breeders of laying hens, gives a consistent 72 to 80 weeks of age, that is, a laying period of 54 to 62 weeks.
So our adopted 62 weeks is conservative for commercial hens in China.
Adopting a laying period of 54 weeks for China would increase its annual number of killed hens by about 15%, and the global number by about 6%.

### Does FAOSTAT measure hens at a point in time, or the cumulative number of hens?

As explained in Section ***, the QCL definitions and standards' glossary includes the following explanation for the term "Egg production":
> Covers all domestic birds which have contributed to egg production during the year, wherever they lay and the corresponding total production, including eggs intended to be used for hatching but excluding waste on farms.

This sentence has two possible interpretations, with crucially different consequences.
"All domestic birds which have contributed to egg production during the year" can mean:
1. All individual birds that laid eggs during the year: this means that the "Laying" element gives the **cumulative** number of birds that have laid eggs in the year.
2. All kinds of domestic birds that lay eggs: in this case, the sentence says nothing about how birds are counted, and the "Laying" element would give the number of birds that lay eggs **at any given time**.

Note that each interpretation would lead to very different numbers:
Imagine a farm that produces 30,000 eggs a year with 100 hens in lay, and replaces each hen after one year, spreading the replacements evenly over the year.
Under Option 2, the "Laying" element should be 100 (with a yield of 300 eggs per hen).
Under Option 1, "Laying" should report 200 hens (with a yield of 150 eggs per hen): the 100 that were there at the beginning of the year, plus the 100 that replaced them during the year.

Based purely on semantics, Option 1 seems to be the more natural interpretation of FAOSTAT's explanation.
All domestic birds which have contributed to egg production is the cumulative number of hens that were in lay during the year.
This could explain the mismatch between FAOSTAT's reported number of laying hens for China and other sources.
However, as we saw in Section ***, the mismatch has a different origin.

However, Option 2 could be interpreted as a clarification that the data includes all species that lay eggs, e.g. hens, ducks, and pigeons, and not just hens.
Option 2 could also be interpreted as a clarification that the data includes breeder hens (of chickens raised both for eggs and chickens raised for meat), and not just laying hens.
